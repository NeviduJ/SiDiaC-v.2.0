```python
def process_document_from_pdf(project_id: str, location: str, processor_id: str, pdf_path: str, credentials_path: str = None):
    """
    Processes a PDF document using the Google Cloud Document AI processor.

    Args:
        project_id (str): Your Google Cloud project ID.
        location (str): The Google Cloud region (e.g., "us", "eu").
        processor_id (str): The ID of your Document AI processor.
        pdf_path (str): The local file path to the PDF document.
        credentials_path (str, optional): Path to a service account JSON key file.
                                          If None, uses Application Default Credentials.

    Returns:
        str: The extracted text from the processed document.
             Returns None if the PDF file is not found or processing fails.
    """

    # Validate PDF file existence
    if not os.path.exists(pdf_path):
        print(f"Error: PDF file not found at '{pdf_path}'")
        return None

    opts = {}
    if location != "us":
        # Document AI API endpoints are regional
        opts = {"api_endpoint": f"{location}-documentai.googleapis.com"}

    # --- START MODIFICATION ---
    if credentials_path:
        if not os.path.exists(credentials_path):
            print(f"Error: Credentials file not found at '{credentials_path}'")
            return None
        # Create credentials object from the service account key file
        credentials = service_account.Credentials.from_service_account_file(credentials_path)
        client = documentai.DocumentProcessorServiceClient(client_options=opts, credentials=credentials)
    else:
        # Fallback to existing logic (Application Default Credentials)
        client = documentai.DocumentProcessorServiceClient(client_options=opts)
    # --- END MODIFICATION ---

    # The full resource name of the processor
    name = client.processor_path(project_id, location, processor_id)

    # Read the PDF file content into bytes
    with open(pdf_path, "rb") as image:
        image_content = image.read()

    raw_document = documentai.RawDocument(
        content=image_content,
        mime_type="application/pdf"
    )

    request = documentai.ProcessRequest(name=name, raw_document=raw_document)

    try:
        result = client.process_document(request=request)
        return result.document
    except Exception as e:
        print(f"Error processing document: {e}")
        return None
```

```python
def get_files_in_directory_os(directory_path):
    """
    Gets a list of all file names (excluding directories) in the specified directory.

    Args:
        directory_path (str): The path to the directory.

    Returns:
        list: A list of strings, where each string is the name of a file.
              Returns an empty list if the directory does not exist or is empty.
    """
    file_names = []
    try:
        # Get all entries (files and directories)
        all_entries = os.listdir(directory_path)

        # Filter to include only files
        for entry in all_entries:
            full_path = os.path.join(directory_path, entry)
            if os.path.isfile(full_path):
                file_names.append(entry)
    except FileNotFoundError:
        print(f"Error: Directory not found at '{directory_path}'")
    except PermissionError:
        print(f"Error: Permission denied to access '{directory_path}'")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

    return file_names
```

```python
output_text_file_path = 'data/OCR_Final/'
error_list = []
model_condfidences = {}
for i in tqdm(range(len(file_list))):

    try:
    
        pdf_file_path = os.path.join('Diachronic Corpus', file_list[i])

        generated_text = process_document_from_pdf(
            project_id=project_id_new,
            location=location_new,
            processor_id=processor_id_new,
            pdf_path=pdf_file_path,
            credentials_path=service_account_key_path # This is the new parameter
        )

        output_file_path = os.path.join(output_text_file_path, file_list[i][:-4] + '.txt')

        os.makedirs(os.path.dirname(output_text_file_path), exist_ok=True)
        with open(output_file_path, "w", encoding="utf-8") as f:
            f.write(generated_text.text)

        confidence = 0
        page_count = 0
        if generated_text.pages:
            for j, page in enumerate(generated_text.pages):
                page_count += 1
                # print(f"Page {i + 1}:")
                if page.detected_languages:
                    for lang in page.detected_languages:
                        confidence += lang.confidence
                        # print(f"Confidence: {lang.confidence:.4f}")
                else:
                    print("No languages detected for this page.")
        else:
            print(f"No pages found in the {file_list[i]}.")

        avg_confidence = confidence/page_count
        # print(avg_confidence)
            
        model_condfidences[file_list[i][:-4]] = avg_confidence

        # if i > 3:
        #     break
    except:
        print(f"Error in document: {file_list[i]}" )
        error_list.append(file_list[i])
```
